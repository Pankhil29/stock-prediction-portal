## 4. Data Preprocessing install scikit_learn,keras,tensorflow

```python
# spliting data into 2 parts 
# 1. 70% of data for training
# 2. 30% of data for testing
data_training = pd.DataFrame(df.Close[0:int(len(df)*0.7)])
data_testing = pd.DataFrame(df.Close[int(len(df)*0.7):len(df)])
print(data_training)
print(data_testing)
```

```python
type(data_training)
```

## Scaling training data into 0 and 1

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(0,1))
```

```python
data_training_array = scaler.fit_transform(data_training)
data_training_array
```

```python
type(data_training_array)
```

# Sequence Creation

```python
# 5 step data sequence
# 2,3,4,5,6 model predict 6th day price is 7 
# 3,4,5,6,7 model predict 7th day price is 8 here 2 is cancel and 7 will be added in dataset.
```

```python
x_train =[]
y_train = []

for i in range(100,  data_training_array.shape[0]):
    x_train.append(data_training_array[i-100:i])
    y_train.append(data_training_array[i,0])
x_train, y_train = np.array(x_train), np.array(y_train)
```

```python
x_train
```

```python
x_train.ndim
```

```python
y_train
```

```python
y_train.ndim
```

# 6 Model Building 

```python
import tensorflow as tf
print(tf.__version__)
```

```python
# ML model 
from keras.models import Sequential
from keras.layers import Dense,LSTM,Input
```

```python
model = Sequential()

model.add(Input(shape=(100,1))) # in this we give the i/p to the model that have 100 days previous data and 1 feature
model.add(LSTM(units=128,activation="tanh",return_sequences=True))
# 128 units = 128 nuerons, activation="tanh" means data ko -1 se 1 ke bich me rkhta hai, je o/p 100days nu analysis karine 100 o/p return krse sequence jethi LSTM layer ne poora sequence made 
# 1st dimaag(Expert) kemke 128 neurons che 
model.add(LSTM(units=64))
# 2nd dimmag(manager) aa je apne apya uper 1st dimag ne ana report ne bariki thi check karse ama return_seq. nathi kemk 100 days nu nichod karine aek chellu result apse
model.add(Dense(25))
# intermediate layer LSTM ae je pattern find karyu aene comprees karine 25 points ma convert kare che
model.add(Dense(1))
# apne end of the output single value joie aetle 25 points ne convert karine aek output ape 
```

# 7.Model Triaing 


```python
model.compile(optimizer="adam",loss="mean_squared_error")
# we give the rules for learning. adam is Coach ye bahut tezi se aur sahi tarike se batata hai ki neurons ki values (weights) ko kaise badalna hai taaki result sahi aaye.
# loss ae scorecard hai aa define kare che ke model ketli error kare che jetlo loss ochho aetlo model saro MSE ae error no square karine avg nikale 
model.fit(x_train,y_train,epochs=40)
# fit means learning start 
# Aap model ko "Sawalaat" (100 din ka data) aur unke "Sahi Jawab" (101-ve din ka price) de rahe hain.
# Epochs=50: Iska matlab hai ki model poore data ko 50 baar shuru se aakhri tak padhega.
# Kyun? Ek baar padhne se koi topper nahi banta. Model pehli baar mein galti karega, optimizer use sudharega, phir wo doosri baar padhega... aise hi 50 baar practice karne se wo patterns ko achi tarah yaad kar lega.
```
