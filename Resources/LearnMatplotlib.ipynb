```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Type of Plot

### Line Plot

```python
x = [1,2,3,4,5]
y = [10,20,25,30,40]
```

```python
plt.plot(x,y);
plt.title("simple line plot ");
plt.xlabel("X-axis label");
plt.ylabel("Y-axis label");
plt.grid(True)
```

## Bar Plot

```python
plt.figure(figsize=[10,6])
month = ["Jan", "Feb","Mar","Apr"]
revenue = [100,200,300,400]
plt.bar(month,revenue)
plt.title("Monthly revenue of bar plot ");
plt.xlabel("X-axis label");
plt.ylabel("Y-axis label");
```

## Scatter plot

## Subplots and Legends 

```python
days = [1,2,3,4,5]
original_price = [10,20,25,30,40]
predicated_price = [8,21,25,28,38]
```

```python
plt.plot(days,original_price, "b-o" ,label="Original Price")
plt.plot(days,predicated_price, "r-o" ,label="Predicted Price")
plt.legend()
plt.title("plot with legend")
plt.xlabel("Weekdays")
plt.ylabel("Price")
```

```python
# Subplots - Orginal price
plt.subplot(3,1,1)
plt.plot(days,original_price, 'b-o',label="Original Price")
plt.legend()
plt.title("Stock price over time")
plt.xlabel("Weekdays")
plt.ylabel("Price")

# Subplots - Predicated Price
plt.subplot(3,1,2)
plt.plot(days,predicated_price, 'r-o',label="Predicated Price")
plt.legend()
plt.title("Stock price over time")
plt.xlabel("Weekdays")
plt.ylabel("Price")

# Subplots -  Difference between Original Price and Predicated Price  
plt.subplot(3,1,3)
price_diff = [o - p for o , p in zip(original_price,predicated_price)]
plt.bar(days,price_diff, color='purple',label="Predicated Price")
plt.legend()
plt.title("Difference between Original Price and Predicated Price")
plt.xlabel("Weekdays")
plt.ylabel("Price")
plt.tight_layout()
```

```python
plt.plot(days,original_price, "b-o" ,label="Original Price")
plt.plot(days,predicated_price, "r-o" ,label="Predicted Price")
plt.legend()
plt.title("plot with legend")
plt.xlabel("Weekdays")
plt.ylabel("Price")

plt.savefig("example_predication.png",dpi=300)
```
